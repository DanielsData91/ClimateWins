#### Prepare Data for Deep Learning

```python
# Drop Columns from the original without enough pertinent data

df_weather_drop = df_weather.drop(columns=[col for col in df_weather.columns if col.startswith(("GDANSK", "TOURS", "ROMA"))])
```

```python
# Check on the new dataframe

df_weather_drop.shape
```

```python
# Trying to find out all the different measurement types for each location

# Extract location names 
locations = set([col.split('_')[0] for col in df_weather_drop.columns])

# Create a dictionary to store measurement counts for each location
measurement_counts = {location: {} for location in locations}

# Count occurrences of each measurement type for each location
for col in df_weather_drop.columns:
    parts = col.split('_') 
    location = parts[0] 
    measurement = '_'.join(parts[1:])  # Join remaining parts if there are more than two

    if measurement not in measurement_counts[location]:
        measurement_counts[location][measurement] = 1
    else:
        measurement_counts[location][measurement] += 1

# Print the measurement counts for each location
for location, measurements in measurement_counts.items():
    print(f"Location: {location}")
    for measurement, count in measurements.items():
        print(f"  - {measurement}: {count}")
    print()
```

__Missing Values:__ Most missing values per weather stations seems to be the snow and wind speed variable. Therefore, I am going to drop them from the data set for further analysis.

```python
# Drop columns for wind_speed and snow_depth measurements since they are missing so often

# Create a list of columns to drop
cols_to_drop = [col for col in df_weather_drop.columns if col.endswith(('wind_speed', 'snow_depth'))]

# Dropping
df_weather_clean = df_weather_drop.drop(columns=cols_to_drop)
```

```python
df_weather_clean.shape
```

```python
# Define relationships between locations
location_pairs = {
    'KASSEL': 'LJUBLJANA',
    'STOCKHOLM': 'OSLO',
    'MUNCHENB': 'SONNBLICK'
}

# Define the desired order of measurements
measurement_order = ['cloud_cover', 'humidity', 'pressure', 'global_radiation', 
                     'precipitation', 'sunshine', 'temp_mean', 'temp_min', 'temp_max']

# Function to fill missing values and insert in correct position
def fill_missing_values(df_weather_clean, location, measurement, neighbor):
    """
    Fills missing values for a given location and measurement using data from a neighboring location.
    Inserts the new column in the correct position based on the measurement order.
    """
    source_col = f'{neighbor}_{measurement}'
    target_col = f'{location}_{measurement}'

    # Determine the correct position to insert the new column
    if measurement == measurement_order[0]:
        location_columns = [col for col in df_weather_clean.columns if col.startswith(location)]
        insert_index = df_weather_clean.columns.get_loc(location_columns[0]) if location_columns else 0
    else:
        prev_measurement = measurement_order[measurement_order.index(measurement) - 1]
        insert_index = df_weather_clean.columns.get_loc(f'{location}_{prev_measurement}') + 1

    # Insert new column with NaN values
    df_weather_clean.insert(insert_index, target_col, np.nan)

    # Fill missing values from the neighbor column
    df_weather_clean[target_col] = df_weather_clean[target_col].fillna(df_weather_clean[source_col])

    return df_weather_clean

# Apply the function to fill missing measurements
for location, neighbor in location_pairs.items():
    for measurement in measurement_order:
        if f'{location}_{measurement}' not in df_weather_clean.columns:
            df_weather_clean = fill_missing_values(df_weather_clean, location, measurement, neighbor)

# Display a summary of newly added columns
selected_columns = [col for col in df_weather_clean.columns if col.startswith(('KASSEL', 'STOCKHOLM', 'MUNCHENB'))]

# Safe output: show only first few columns and rows to avoid crashing notebook
print("Sample of newly added columns:")
print(df_weather_clean[selected_columns[:5]].head())
print("\nTotal new columns added:", len(selected_columns))
```

# 5. Create a Keras Model for RNN Algorithm and Compile & Run the RNN Model

## 5.1 First Approach

```python
epochs = 20
batch_size = 16
n_hidden = 16

timesteps = len(X_train[0])
input_dim = len(X_train[0][0])
n_classes = len(y_train[0])

model = Sequential()
model.add(LSTM(n_hidden, input_shape=(timesteps, input_dim)))
model.add(Dropout(0.5))
model.add(Dense(n_classes, activation='sigmoid'))
```

```python
model.summary()
```

### First Run

```python
model.compile(loss='categorical_crossentropy',
              optimizer='rmsprop',
              metrics=['accuracy'])
```

```python
model.fit(X_train,
          y_train,
          batch_size=batch_size,
          validation_data=(X_test, y_test),
          epochs=epochs)
```

### Confusion Matrix

```python
# Define list of stations names
stations = {
0: 'BASEL',
1: 'BELGRADE',
2: 'BUDAPEST',
3: 'DEBILT',
4: 'DUSSELDORF',
5: 'HEATHROW',
6: 'KASSEL',
7: 'LJUBLJANA',
8: 'MAASTRICHT',
9: 'MADRID',
10: 'MUNCHENB',
11: 'OSLO',
12: 'SONNBLICK',
13: 'STOCKHOLM',
14: 'VALENTIA'
}
```

```python
def confusion_matrix(y_true, y_pred):
    y_true = pd.Series([stations[y] for y in np.argmax(y_true, axis=1)])
    y_pred = pd.Series([stations[y] for y in np.argmax(y_pred, axis=1)])

    return pd.crosstab(y_true, y_pred, rownames=['True'], colnames=['Pred'])
```

```python
# Evaluate
print(confusion_matrix(y_test, model.predict(X_test)))
```

## 5.2 Second Approach

```python
# Key parameters
epochs = 20
batch_size = 32 # --> INCREASED from 16 to 32
n_hidden = 32  # --> INCREASED from 16 to 32
learning_rate = 0.0005 # --> ADDED Learning Rate

# Shape info
timesteps = len(X_train[0])
input_dim = len(X_train[0][0])
n_classes = len(y_train[0])


model = Sequential()
model.add(Conv1D(filters=64, kernel_size=3, activation='relu', padding='same', 
                 input_shape=(timesteps, input_dim))) # --> ADDED Convolutional Layer
model.add(MaxPooling1D(pool_size=2)) # --> ADDED Convolutional Pooling

# 2. LSTM Layer (Temporal modeling) ## --> ADDED
model.add(LSTM(n_hidden, return_sequences=False))
model.add(Dropout(0.5))

# 3. Dense Output Layer
# Use softmax for multi-class classification
model.add(Dense(n_classes, activation='sigmoid'))
# Summary
model.summary()
```
